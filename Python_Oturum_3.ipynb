TRY - EXCEPT - FINALLY


```python

try:
  a = float(input("Bölünen"))
  b = float(input("Bölen"))
  print(a/b)
except ValueError:
  print("Lütfen geçerli bir sayı giriniz.")
except ZeroDivisionError:
  print("Bölen 0 olursa sonsuza gidecektir. Matemetiksel olarak bir şey temsil etmez")
except KeyboardInterrupt:
  print("Çıkış yapılıyor.")
except:
  print("Bu bir hatadır.")
finally:
  print("Tech Istanbul Bootcamp")
```
