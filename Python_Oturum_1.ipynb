```python
sayi_1 = float(input("Sayı 1 : "))
sembol = input("Yapılacak işlem:")
sayi_2= float(input("Sayı 2 :"))


if sembol =="+":
 print(sayi_1+sayi_2)
elif sembol == '-':
  print(sayi_1-sayi_2)
elif sembol == '*':
  print(sayi_1*sayi_2)
elif sembol == '/':
  print(sayi_1/sayi_2)
else:
  print("Hatalı sembol girişi.")

```

```python
ogrenci_notu=float(input("Lütfen notunuzu giriniz : "))

if ogrenci_notu >= 85:
  print("Tebrikler! Takdir Belgesi aldınız.")
elif ogrenci_notu >=70:
  print("Tebrikler! Teşekkürler Belgesi aldınız.")
elif ogrenci_notu < 40:
  print ("Sınıfta kaldınız.")
else:
  print("Sınıfı belgesiz geçtiniz.")
```

```python
ogrenci_notu=float(input("Lütfen notunuzu giriniz : "))

if ogrenci_notu >= 85 and ogrenci_notu <=100:
  print("Tebrikler! Takdir Belgesi aldınız.")
elif ogrenci_notu >=70 and ogrenci_notu <85:
  print("Tebrikler! Teşekkürler Belgesi aldınız.")
elif ogrenci_notu < 70 and ogrenci_notu >=45:
  print ("Sınıfı belgesiz geçtiniz")
elif ogrenci_notu < 45 and ogrenci_notu >=0:
  print ("Sınıfta kaldınız.")
else:
  print("Hatalı değer girdiniz.")
```
